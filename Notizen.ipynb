### camera_reader_node

ROS-Knoten starten
Der Code startet einen ROS-Knoten namens camera_reader_node.

Bild von der Duckiebot-Kamera empfangen
Das Bild kommt über das ROS-Topic /[VEHICLE_NAME]/camera_node/image/compressed.

Bild umwandeln und anzeigen

Das Bild wird in ein OpenCV-Format umgewandelt.

Je nach Modus wird das Bild farblich gefiltert oder es werden Linien eingezeichnet.

Konfigurierbare Parameter laden
Aus einer YAML-Datei werden vordefinierte Parameter (Farbbereiche, Eckpunkte etc.) geladen.

Tkinter GUI anzeigen
Eine Benutzeroberfläche mit:

Dropdown-Menü (zum Auswählen des Modus, z. B. „gelbe Linie“, „blaue Linie“, „lane_image“),

Slidern (um Farbfilter oder Koordinaten anzupassen),

Live-Bildanzeige (das gefilterte oder markierte Kamerabild).

Parameter anpassen & Live-Auswirkung sehen
Du kannst die Werte mit Slidern verändern und siehst sofort im Bild, wie sich das auswirkt.

Beim Beenden: Parameter ausgeben
Wenn du das Fenster schließt, werden die aktuellen Einstellungen im Terminal ausgegeben (als YAML-Text).

### control_lane_node


ROS-Knoten starten:
Ein Knoten namens control_lane_node wird erstellt.

Eingehende Daten:

/detect/lane → Gibt an, wie weit die Spur vom Bildzentrum entfernt ist (Pixelkoordinate).

/switch/control → Aktiviert/Deaktiviert den Lane-Follow-Modus (je nach ausgewähltem Steuermodus, z. B. manuell vs. automatisch).

Lane-Following-Logik:

Wenn aktiviert (self.enable == True), wird bei jedem neuen lane-Wert ein Steuerbefehl berechnet:

Fehler = Abstand zur Bildmitte (500 px) → normalisiert durch 100.

v = konstante Geschwindigkeit = 0.2

omega = proportionaler Fehler → einfache P-Regelung

Steuerbefehl senden:
Der berechnete Befehl (v, omega) wird an den Duckiebot gesendet über das Topic /car_cmd_switch_node/cmd.

Sicheres Abschalten:
Beim Beenden wird ein Stopp-Befehl (v=0, omega=0) gesendet.

### control_obstacle_node

ROS-Knoten starten:
Der Node control_obstacle_node wird erstellt.

Wichtige Subscriptions:

/detect/duckie: Gibt z. B. den Abstand oder die Position eines Hindernisses an.

/switch/control: Schaltet den Hindernisvermeidungsmodus ein oder aus, je nach Steuermodus.

Logik:

Wenn der Modus Obstacle aktiviert ist (enable == True), wird bei einer Duckie-Erkennung die Callback cbAvoideObstacle() ausgeführt.

Diese Funktion ist noch leer – hier soll später deine Hindernisvermeidung implementiert werden (z. B. stoppen, ausweichen, umfahren).

Steuerbefehle:

Steuerbefehle werden über /car_cmd_switch_node/cmd publiziert.

Beim Beenden wird der Bot gestoppt (v=0, omega=0).

### detect_duckie_node


Initialisierung:

Das YOLOv8-Modell (model.pt) wird geladen.

Die Kamera (/image/compressed) wird abonniert.

Zwei Publisher:

Annotiertes Bild → /detect/duckie/image

Float-Wert zur Duckie-Erkennung → /detect/duckie (noch nicht implementiert)

Objekterkennung (cbDetectObjects)

Alle 3 Bilder (zur Reduktion der Rechenlast) werden verarbeitet.

YOLO erkennt Objekte im Bild.

Bounding Boxes und Klassenbezeichnungen werden eingezeichnet.

Das Ergebnisbild wird als ROS-Image veröffentlicht.

(Noch nicht implementiert):

Es wird kein konkreter Float64-Wert (z. B. Duckie-Abstand) auf /detect/duckie gesendet – nur das Bild.
